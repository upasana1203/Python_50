```python
class LargestElementFinder:

    def __init__(self, numbers):
        self.numbers = numbers

    def find_largest(self):
        # -------- Input Validation --------
        if not isinstance(self.numbers, list):
            raise TypeError("Input must be a list")

        if len(self.numbers) == 0:
            raise ValueError("List must not be empty")

        # Check if all elements are numbers
        for num in self.numbers:
            if not isinstance(num, (int, float)):
                raise TypeError("List must contain only numeric values")

        # -------- Logic using for loop --------
        largest = self.numbers[0]
        all_same = True

        for num in self.numbers:
            if num > largest:
                largest = num
            if num != self.numbers[0]:
                all_same = False

        # -------- Final Decision --------
        if all_same:
            return None

        return largest
        
if __name__ == "__main__":
    try:
        numbers = 
        finder = LargestElementFinder([7, 5, 8, 2, 10, 9])
        print(finder.find_largest())  # Output: 10

        finder2 = LargestElementFinder([4, 4, 4, 4])
        print(finder2.find_largest())  # Output: None

    except (TypeError, ValueError) as e:
        print("Error:", e)
```
