```python
class Fibbonacci:
    def __init__(self , num):
        if not isinstance(num , int):
            raise TypeError("TypeError : input must be int")

        if num <= 0:
            raise ValueError("ValueError : input should be non- negative integer")

        self.num = num

    def fibbonacci_series(self):
        if self.num <= 1:
            return 1

        a, b = 0 ,1
        for _ in range(self.num):
            print(a , end = " ")
            a , b = b , a+b
             

if __name__ == "__main__":
    try:
        user_num = Fibbonacci(5)
        user_num.fibbonacci_series()
#         # print(result)

    except Exception as e:
        print(e)
```
