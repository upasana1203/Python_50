```python
class Intersection:
    def __init__(self , list1 , list2):
        if not isinstance(list1 , list):
            raise TypeError("INPUT MUST BE LIST")

        if not isinstance(list2 , list):
            raise TypeError("INPUT MUST BE LIST")

        if len(list1) <= 1 or len(list2) <= 1 :
            raise ValueError("input must have 2 elements")
            
        self.list1 = list1
        self.list2 = list2

    def list_intersection(self):
        common = []
        for i in self.list1:
            if i in self.list2:
                common.append(i)
            else:
                return f"no common elements"
                              
        return common

if __name__ == "__main__":
    try:
        obj = Intersection([1,2] , [3 ,4])
        print(obj.list_intersection())

    except Exception as e:
        print(e)
```
