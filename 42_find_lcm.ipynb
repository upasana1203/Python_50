```python
class Find_lcm:
    def __init__(self , a , b):
        if not isinstance(a , (int,float)) or not isinstance(b,(int , float)):
            raise TypeError("input must be either int of float type")

        if not a or not b:
            raise ValueError("input should not be empty")
        self.a = a
        self.b = b
        
    def LCM(self):
        greater = max(self.a, self.b)
        smallest = min(self.a, self.b)
        for i in range(greater, self.a*self.b+1, greater):
            if i % smallest == 0:
                return i


if __name__ == '__main__':
    try:
        # a = int(input("enter first num : "))
        # b = int(input("enter first num : "))
        obj = Find_lcm(12 , 9 )
        result = obj.LCM()
        print("LCM of", a, "and", b, "is", LCM(a, b))
    except Exception as e:
        print(e)
```
