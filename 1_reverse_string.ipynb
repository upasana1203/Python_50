1. REVERSE STRING

```python
class Reverse:
    def __init__(self,user_str):
        if not isinstance(user_str, str):
            raise TypeError("Input must be of type 'str'")

        if user_str == "":
            raise ValueError("Input should not be empty")

        if user_str.strip() == "":
            raise ValueError("Input should not contain only spaces")
        self.user_str = user_str

    def reverse_string(self):
        reverse = ""
        for i in self.user_str:
            reverse = i + reverse
        return reverse

if __name__ == "__main__":
    try:
        user_input = input("enter your input : ")
        reverse_obj = Reverse(user_input)
        print(reverse_obj.reverse_string())

    except Exception as e:
        print(e)
```

```python
def reverse(word):
    reverse_str = word[::-1]
    return reverse_str

reverse("study please")
```

```python
# by recursion
def reverse_str(s):
    if s == "":
        return s
    return reverse_str(s[1:]) + s[0]

reverse_str("prove")
```

```python
def reverse_str(s):
    stack = []

    for ch in s:
        stack.append(ch)

    result = ""
    while stack:
        result += stack.pop()

    return result
reverse_str("nowww")
```

```python
from functools import reduce

def reverse_str(s):
    return reduce(lambda x, y: y + x, s)

reverse_str("xyzlambda")
```
