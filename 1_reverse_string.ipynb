1. REVERSE STRING

```python
def reverse_str(input_str: str) -> str:
    """
    Reverses a given string.

    Raises:
        TypeError: If input is not a string
        ValueError: If input is empty, None, or contains only spaces
    """

    if input_str is None:
        raise ValueError("Input cannot be None")

    if not isinstance(input_str, str):
        raise TypeError("Input must be of type 'str'")

    if input_str == "":
        raise ValueError("Input should not be empty")

    if input_str.strip() == "":
        raise ValueError("Input should not contain only spaces")

    reverse_string = ""
    for ch in input_str:
        reverse_string = ch + reverse_string

    return reverse_string
if __name__ == "__main__":
    try:
        user_input = input("Enter your string: ")
        result = reverse_str(user_input)
        print(f"The reversed string is: {result}")

    except (TypeError, ValueError) as e:
        print(f"Error: {e}")
```

```python
def reverse(word):
    reverse_str = word[::-1]
    return reverse_str

reverse("study please")
```

```python
# by recursion
def reverse_str(s):
    if s == "":
        return s
    return reverse_str(s[1:]) + s[0]

reverse_str("prove")
```

```python
def reverse_str(s):
    stack = []

    for ch in s:
        stack.append(ch)

    result = ""
    while stack:
        result += stack.pop()

    return result
reverse_str("nowww")
```

```python
from functools import reduce

def reverse_str(s):
    return reduce(lambda x, y: y + x, s)

reverse_str("xyzlambda")
```
