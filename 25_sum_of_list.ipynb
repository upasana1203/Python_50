```python
class SumOfList:
    def __init__(self , list1):
        if not isinstance(list1 , list):
            raise TypeError("TypeError : the input must be list type")

        if len(list1) == 0:
            raise ValueError("ValeError : input should not be empty")
            
        self.list = list1

    def sum_of_list(self):
        sum = 0
        for i in self.list:
            if not isinstance(i , (int , float)):
                raise TypeError("ValueError : List elements must be int or float")
            
            sum += i

        return sum

if __name__ == "__main__":
    try:
        obj = SumOfList([10 , 20 , 30])
        print(obj.sum_of_list())
    except Exception as e:
        print (e)
```
