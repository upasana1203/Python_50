```python
class Anagram:
    def __init__(self , input1 , input2):
        if not isinstance(input1 , str) or not isinstance(input2 , str):
            raise TypeError("TypeError : input must be string")

        if input1.strip() == "":
            raise ValueError("valueError : input1 is empty")

        if input2.strip() == "":
            raise ValueError("valueError : input2 is empty")

        input1 = input1.replace(" ","").lower()
        input2 = input2.replace(" ","").lower()
        
        self.input1 = input1
        self.input2 = input2

    def is_anagram(self):

        
        if len(self.input1) != len(self.input2):
            return False

        count = {}

        for i in self.input1:
            count[i] = count.get(i , 0) + 1

        for i in self.input2:
            if i not in count or count[i] == 0:
                return False
            count[i] -= 1
        return True

if __name__ == "__main__":
    try:
        user_input = Anagram("mom" , "Mom")
        print(user_input.is_anagram())

    except Exception as e:
        print(e)
```
