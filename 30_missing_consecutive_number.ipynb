```python
class Consecutive:
    def __init__(self, lst):
        if not isinstance(lst, list):
            raise TypeError("Input must be a list")

        if len(lst) < 2:
            raise ValueError("Input must contain at least two elements")

        if any(isinstance(x, bool) for x in lst):
            raise TypeError("Boolean values are not allowed")

        if not all(isinstance(x, int) for x in lst):
            raise TypeError("All elements must be integers")

        if lst != sorted(lst):
            raise ValueError("List must be sorted in ascending order")

        self.lst = lst

    def missing_consecutive(self):
        gaps = 0
        missing = None

        for i in range(len(self.lst) - 1):
            diff = self.lst[i + 1] - self.lst[i]

            if diff == 1:
                continue
            elif diff == 2:
                missing = self.lst[i] + 1
                gaps += 1
            else:
                raise ValueError("Invalid sequence: gap larger than 1")

        if gaps == 0:
            raise ValueError("No missing number found")

        if gaps > 1:
            raise ValueError("More than one missing number found")

        return missing


if __name__ == "__main__":
    try:
        user_obj = Consecutive([1, 3])
        print(user_obj.missing_consecutive())
    except Exception as e:
        print(e)
```
