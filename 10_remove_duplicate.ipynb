```python
class Duplicate:
    def __init__(self , user_list):
        if not isinstance(user_list , list):
            raise TypeError("TypeError : input must be list")

        if len(user_list) == 0:
            raise ValueError ("ValueError : input can not be empty")
            
        self.list = user_list

    def remove_duplicate(self):
        unique = []

        for i in self.list:
            if i not in unique:
                unique.append(i)
        return unique

if __name__ == "__main__":
    try:
        user_input = Duplicate([1,3,4,"this",3,2,6])
        result = user_input.remove_duplicate()
        print(result)
    except Exception as e:
        print(e)
```
