```python
class LeapYear:
    def __init__(self , input_year):
        if not isinstance(input_year , int):
            raise TypeError("TypeError : input must be 'int'")
        if input_year <= 0:
            raise ValueError("ValueError : year must be greater than 0")

        self.year = input_year

    def is_leap_year(self):
        if (self.year % 4 == 0 and self.year % 100 != 0) or self.year% 400 == 0:
            return f"{self.year} is a leap year"
        else:
            return f"{self.year} is not a leap year"

if __name__ == "__main__":
    try:
        user_year = int(input("Enter your year to check : "))
        year_obj = LeapYear(user_year)
        print(year_obj.is_leap_year())

    except Exception as e:
        print(e)
    
```
