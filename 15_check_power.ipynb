```python
class Power:
    def __init__(self, x, y):
        if not isinstance(x, (int, float)):
            raise TypeError("TypeError: x must be a number")

        if not isinstance(y, int):
            raise TypeError("TypeError: y must be an integer")

        if y < 0:
            raise ValueError("ValueError: negative power not supported")

        self.x = x
        self.y = y

    def calculate_power(self):
        result = 1
        for _ in range(self.y):
            result *= self.x
        return f"result is : {result}"

if __name__ == "__main__":
    try:
        user_num = int(input("enter number : "))
        user_power = int(input("enter number for power : "))
        power_obj = Power(user_num, user_power)
        print(power_obj.calculate_power())

    except Exception as e:
        print(e)
```
