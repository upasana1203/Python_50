```python
class PossiblePairs:
    def __init__(self , lst , tar):
        self.lst = lst
        self.tar = tar

    def possible(self):
        result = []
        for i in range(len(self.lst)):
            for j in range(i+1 , len(self.lst)):
                if self.lst[i] + self.lst[j] == self.tar:
                    result.append((self.lst[i] ,self.lst[j]))

        return result

if __name__ == "__main__":
    try:
        obj = PossiblePairs([1, 2, 3, 4, 5, 6] , 7)
        print(obj.possible())

    except Exception as e:
        print(e)
                    
```

```python
class PossiblePairs:
    def __init__(self , lst , tar):
        self.lst = lst
        self.tar = tar

    def possible(self):
        already_looked = set()
        possible_pairs = []

        for i in self.lst:
            complement = self.tar - i

        if complement in already_looked:
            possible_pairs.append((complement , i))

        already_looked.add(i)

        return possible_pairs

if __name__ == "__main__":
    try:
        obj = PossiblePairs([1, 2, 3, 4, 5, 6] , 7)
        print(obj.possible())

    except Exception as e:
        print(e)
                    
```
