```python
class SortedList:
    def __init__(self, user_lst):
        """
        Initializes the SortedList object.

        Args:
            user_lst (list): List to be checked.

        Raises:
            TypeError: If input is not a list or contains invalid types.
            ValueError: If list contains fewer than 2 elements.
        """
        if not isinstance(user_lst, list):
            raise TypeError("Input must be a list")

        if len(user_lst) < 2:
            raise ValueError("Input must contain at least 2 elements")

        if any(isinstance(x, bool) for x in user_lst):
            raise TypeError("Boolean values are not allowed")

        if not all(isinstance(x, (int, float)) for x in user_lst):
            raise TypeError("All elements must be int or float")

        self.list = user_lst

    def is_sorted(self):
        """
        Checks if the list is sorted in ascending order.

        Returns:
            bool: True if sorted, False otherwise.
        """
        for i in range(len(self.list) - 1):
            if self.list[i] > self.list[i + 1]:
                return False
        return True


if __name__ == "__main__":
    try:
        obj = SortedList([1, 2, 2, 3, 4])
        print(obj.is_sorted())
    except Exception as e:
        print(e)
```
