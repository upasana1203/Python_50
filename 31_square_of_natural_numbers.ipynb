```python
class SquareOfNatural:
    def __init__(self, num):
        if isinstance(num, bool):
            raise TypeError("Boolean values are not allowed")

        if not isinstance(num, int):
            raise TypeError("Input must be an integer")

        if num < 1:
            raise ValueError("Input must be a positive integer (>= 1)")

        self.num = num

    def square_natural_num(self):
        total = 0
        for i in range(1, self.num + 1):
            total += i ** 2
        return f"The sum of squares of first {self.num} natural numbers is: {total}"


if __name__ == "__main__":
    try:
        obj = SquareOfNatural(6)
        print(obj.square_natural_num())
    except Exception as e:
        print(e)
```
