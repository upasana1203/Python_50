```python
class GeneratorOfEvenNumbers:
    def __init__(self , user_range):
        if not isinstance(user_range , int):
            raise TypeError("TypeError : input must be integer")

        if user_range == 0:
            raise ValueError("ValueError : input should not be 0")

        if user_range < 0:
            raise ValueError("ValueError : input should be non- negative integer")
        self.range = user_range

    def generate_even_num(self):
        even_list = []
        for i in range (1,self.range):
            if i % 2 == 0:
                even_list.append(i)
        return even_list

if __name__ == "__main__":
    try:
        input_range = int(input("enter the range to generate even numbers : "))
        generate_obj = GeneratorOfEvenNumbers(input_range)
        print(generate_obj.generate_even_num())

    except Exception as e:
        print(e)
```
