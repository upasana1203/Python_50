```python
class DigitSum:
    def __init__(self , num):
        if not isinstance(num , int):
            raise TypeError("TypeError : input must be int")

        if num < 0:
            raise ValueError("ValueError : input must be non-negative integer")
        self.num = num

    def sum_of_digit(self):
        sum = 0
        while self.num > 0:
            digit = self.num % 10
            sum += digit
            self.num //= 10
        return sum

if __name__ == "__main__":
    
    try :
        user_num = DigitSum(1234)
        result = user_num.sum_of_digit()
        print(result)

    except Exception as e:
        print(e)
```

```python
def sum_of_digits(num):
    if not isinstance(num, int):
        raise TypeError("Input must be an integer")

    if num < 0:
        raise ValueError("Input must be non-negative")

    total = 0
    for digit in str(num):
        total += int(digit)

    return total
```
