```python
class DigitChecker:
    def __init__(self, string):
        self.string = string

    def is_digit_only(self):
        if not self.string:
            return False

        for ch in self.string:
            if ch < '0' or ch > '9':
                return False
        return True

if __name__ == "__main__":
    try:
        obj = DigitChecker("5677bj8")
        print("Does the input contains only num : " ,obj.is_digit_only())

    except Exception as e:
        print(e)
```
