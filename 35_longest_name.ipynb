```python
class FindLongest:
    def __init__(self, sentence):
        if isinstance(sentence, bool):
            raise TypeError("Boolean values are not allowed")

        if not isinstance(sentence, str):
            raise TypeError("Input must be a string")

        if not sentence.strip():
            raise ValueError("Sentence must not be empty")

        self.sent = sentence

    def find_longest(self):
        words = self.sent.split()
        return max(words, key=len)

if __name__ == "__main__":
    try:
        obj = FindLongest("i got supercalifragilistic score in last test")
        print("the longest word is of length : ", obj.find_longest())
    except Exception as e:
        print(e)
```
