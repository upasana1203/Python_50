```python
class Armstrong:
    def __init__(self, number):
        if not isinstance(number, str):
            raise TypeError("Input must be string")

        if not number.isdigit():
            raise ValueError("Input must contain only digits")

        self.number = number

    def is_armstrong(self):
        num_len = len(self.number)
        total = 0

        for digit in self.number:
            total += int(digit) ** num_len

        return total == int(self.number)


if __name__ == "__main__":
    try:
        num_input = input("Enter number: ")
        obj = Armstrong(num_input)
        print("Is Armstrong Number:", obj.is_armstrong())

    except Exception as e:
        print("Error:", e)
```

```python
class Armstrong:
    def __init__(self, num):
        if not isinstance(num, int) or num < 0:
            raise ValueError("Input must be non-negative integer")
        self.num = num

    def is_armstrong(self):
        temp = self.num
        digits = len(str(self.num))
        total = 0

        while temp > 0:
            digit = temp % 10
            total += digit ** digits
            temp //= 10

        return total == self.num

if __name__ == "__main__":
    try:
        num_input = input("Enter number: ")
        obj = Armstrong(num_input)
        print("Is Armstrong Number:", obj.is_armstrong())

    except Exception as e:
        print("Error:", e)
        
```
