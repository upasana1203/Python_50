```python
class Factorial:

    def __init__(self, num):
        self.num = num

    def factorial(self):
        """
        Function:
            Calculates factorial of a number (5! = 5*4*3*2*1)

        Returns:
            int: factorial of the number

        Raises:
            TypeError: if input is not an integer
            ValueError: if input is negative
        """

        if not isinstance(self.num, int):
            raise TypeError("input must be an integer")

        if self.num < 0:
            raise ValueError("factorial is not defined for negative numbers")

        fact = 1
        for i in range(self.num, 0, -1):
            fact *= i

        return fact


if __name__ == "__main__":
    try:
        num_input = int(input("Enter the number: "))
        fact_obj = Factorial(num_input)
        result = fact_obj.factorial()
        print(f"The factorial of {num_input} is {result}")

    except ValueError as e:
        print("Value Error:", e)

    except TypeError as e:
        print("Type Error:", e)
```
