```python
class Prime:
    """
    A class to find all prime numbers up to a given integer.
    """

    def __init__(self, num):
        if isinstance(num, bool):
            raise TypeError("Boolean value is not allowed")

        if not isinstance(num, int):
            raise TypeError("Input must be an integer")

        if num < 2:
            raise ValueError("Input must be greater than or equal to 2")

        self.num = num

    def find_primes(self):
        primes = []

        for n in range(2, self.num + 1):
            is_prime = True

            for i in range(2, int(n ** 0.5) + 1):
                if n % i == 0:
                    is_prime = False
                    break

            if is_prime:
                primes.append(n)

        return primes


if __name__ == "__main__":
    try:
        obj = Prime(20)
        print(obj.find_primes())
    except Exception as e:
        print(e)
```
