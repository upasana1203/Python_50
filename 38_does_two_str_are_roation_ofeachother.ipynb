```python
class IsRotation:
    def __init__(self, str1, str2):
        if not isinstance(str1, str) or not isinstance(str2, str):
            raise TypeError("Both inputs must be strings")

        if not str1 or not str2:
            raise ValueError("Inputs cannot be empty")

        if len(str1) != len(str2):
            raise ValueError("Strings must have the same length")

        self.str1 = str1
        self.str2 = str2

    def check_rotation(self):
        return self.str2 in (self.str1 + self.str1)


if __name__ == "__main__":
    try:
        obj = IsRotation("abcd", "dabc")
        print(obj.check_rotation())

    except Exception as e:
        print(e)
```

```python
class IsRotation:
    def __init__(self, str1, str2):
        self.str1 = str1
        self.str2 = str2

    def rotate(self):
        k = len(self.str1) - 1
        result = self.str1[k:] + self.str1[:k]
        return result

    def check_rotation(self):
        if self.rotate() == self.str2:
            return True
        else:
            return False

if __name__ == "__main__":
    try:
        obj = IsRotation("fghjkl" , "lfghjk")
        print(obj.check_rotation())
    except Exception as e:
        print(e)
```

```python
def rotate(str):
    k = len(str) - 1
    result = str[k:] + str[:k]
    return result
        
rotate('abcd')
```
