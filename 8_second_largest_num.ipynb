```python
class SecondLargestNum:
    def __init__(self , user_list):
        if not isinstance(user_list , list):
            raise TypeError("TypeError : input must be list")

        if len(user_list) == 0:
            raise ValueError("ValueError : input can not be empty")
        
        self.list = user_list

    def second_largest_value(self):

        if len(self.list) < 2:
            return None
            
        largest = None
        second_largest = None

        for i in self.list:
            if not isinstance(i , (int , float)):
                raise TypeError("TypeError : list must contain only int or float")

            
            if largest is None or i >largest:
                if largest is not None and i != largest:
                    second_largest = largest
                largest = i

            elif i != largest and (second_largest is None or i > second_largest):
                second_largest = i
        return second_largest

if __name__ == "__main__":
    try:
        user_input = SecondLargestNum([8,7,9,56,3,78,-90])
        result = user_input.second_largest_value()
        print(f"the second largest num in input is : {result}")
    
    except (TypeError , ValueError) as e:
        print(e)

             
```
