```python
class CommonElements:
    def __init__(self, lst1, lst2, lst3):
        # if 
        self.lst1 = lst1
        self.lst2 = lst2
        self.lst3 = lst3

    def find_common(self):
        result = []
        for item in self.lst1:
            if item in self.lst2 and item in self.lst3:
                if item not in result:
                    result.append(item)
        return result

if __name__ == "__main__":
    try:
        obj = CommonElements(
            [1, 2, 2, 3, 4, 5],
    [2, 3, 5, 5],
    [2, 5, 6]
        )
        print(obj.find_common())

    except Exception as e:
        print("Error:", e)

    
```
